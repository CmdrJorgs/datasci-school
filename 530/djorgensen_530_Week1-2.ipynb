# Chapter 3, Exercise 1
We want to look at data for the Facebook, Apple, Amazon, Netflix, and Google (FAANG) stocks, but we were given each as a separate CSV file (obtained using the stock_analysis package we will build in Chapter 7, Financial Analysis – Bitcoin and the Stock Market). Combine them into a single file and store the dataframe of the FAANG data as faang for the rest of the exercises:
- Read in the `aapl.csv`, `amzn.csv`, `fb.csv`, `goog.csv`, and `nflx.csv` files.
- Add a column to each dataframe, called ticker, indicating the ticker symbol it is for (Apple's is AAPL, for example); this is how you look up a stock. In this case, the filenames happen to be the ticker symbols.
- Append them together into a single dataframe.
- Save the result in a CSV file called `faang.csv`


```python
# Read in the aapl.csv, amzn.csv, fb.csv, goog.csv, and nflx.csv files
import pandas as pd
aapl = pd.read_csv('textbook/ch_03/exercises/aapl.csv')
amzn = pd.read_csv('textbook/ch_03/exercises/amzn.csv')
fb = pd.read_csv('textbook/ch_03/exercises/fb.csv')
goog = pd.read_csv('textbook/ch_03/exercises/goog.csv')
nflx = pd.read_csv('textbook/ch_03/exercises/nflx.csv')
```

```python
# Add ticker columns to each dataframe
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
fb['ticker'] = 'FB'
goog['ticker'] = 'GOOG'
nflx['ticker'] = 'NFLX'
```

```python
# Append stock tables together into a single dataframe
stocks = pd.concat([aapl, amzn, fb, goog, nflx], ignore_index=True)
```

```python
# Save the result in a CSV file
stocks.to_csv('faang.csv', index=False)
```

With faang, use type conversion to cast the values of the date column into datetimes and the volume column into integers. Then, sort by date and ticker.

```python
# Cast the values of the date column into datetimes and the volume column into integers
stocks['date'] = pd.to_datetime(stocks['date'])
stocks['volume'] = stocks['volume'].astype(int)
```

# Chapter 3, Exercise 6
[The European Centre for Disease Prevention and Control (ECDC)](https://www.ecdc.europa.eu/en/publications-data/download-todays-data-geographic-distribution-covid-19-cases-worldwide) provides an open dataset on COVID-19 cases called daily number of new reported cases of COVID-19 by country worldwide. This dataset is updated daily, but we will use a snapshot that contains data from January 1, 2020 through September 18, 2020. Clean and pivot the data so that it is in wide format:
- Read in the *covid19\_cases.csv* file.
- Create a date column using the data in the *dateRep* column and the *pd.to\_datetime()* function.
- Set the date column as the index and sort the index.
- Replace all occurrences of *United\_States\_of\_America* and *United\_Kingdom* with *USA* and *UK*, respectively. Hint: the replace() method can be run on the dataframe as a whole.
- Using the *countriesAndTerritories* column, filter the cleaned COVID-19 cases data down to Argentina, Brazil, China, Colombia, India, Italy, Mexico, Peru, Russia, Spain, Turkey, the UK, and the USA.
- Pivot the data so that the index contains the dates, the columns contain the country names, and the values are the case counts (the *cases* column). Be sure to fill in NaN values with 0.

```python
# Read in the covid19_cases.csv file
covid19 = pd.read_csv('textbook/ch_03/exercises/covid19_cases.csv')
covid19_total = pd.read_csv('textbook/ch_03/exercises/covid19_total_cases.csv')
```

```python
# Create a date column in covid19
covid19['date'] = pd.to_datetime(covid19['dateRep'], format='%d/%m/%Y')
```

```python
# Set the date column as the index and sort the index
covid19.set_index('date', inplace=True)
covid19 = covid19.sort_index()
```

```python
# Replace all occurrences of United_States_of_America and United_Kingdom with USA and UK
covid19['countriesAndTerritories'] = covid19['countriesAndTerritories'].replace({
    'United_States_of_America': 'USA',
    'United_Kingdom': 'UK'
})
```

```python
# Filter the cleaned COVID-19 cases data down to Argentina, Brazil, China, Colombia, India, Italy, Mexico, Peru, Russia, Spain, Turkey, the UK, and the USA
countries = ['Argentina', 'Brazil', 'China', 'Colombia', 'India', 'Italy', 'Mexico', 'Peru', 'Russia', 'Spain', 'Turkey', 'UK', 'USA']
filtered_covid19 = covid19[covid19['countriesAndTerritories'].isin(countries)]
```

```python
# Pivot the data
pivoted_covid19 = filtered_covid19.pivot_table(
    index=filtered_covid19.index, # dates are the index
    columns='countriesAndTerritories', # country names are the columns
    values='cases', # values are from the cases column
    aggfunc='sum', # in case there are multiple entries for the same date and country, sum them
    fill_value=0 # replace NaN values with 0
)
pivoted_covid19
```
