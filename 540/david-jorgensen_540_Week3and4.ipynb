```python
# Load data
income_df = pd.read_csv('./adult_income_data.csv')
# Display the first 10 rows of the dataset
income_df.head(10)
```

```python
# Add missing header names to the dataset by reading ./adult_income_names.txt line by line and assigning the names to the columns in the dataset
with open('./adult_income_names.txt', 'r') as f:
    column_names = f.read().splitlines()
    # remove empty lines from column_names
    column_names = [name for name in column_names if name.strip() != '']
    # split column names by ':' and take the first part as the column name
    column_names = [name.split(':')[0].strip() for name in column_names]
# Add an extra column name called 'name-of-income' to the end of the column names list to account for the response variable column in the dataset
column_names.append('name-of-income')
# print column names to verify they are correct
print("Column names from adult_income_names.txt:")
print(column_names)
# Set the column names in the dataset
income_df.columns = column_names
# Display the first 10 rows of the updated dataset
income_df.head(10)
```

```python
# Find missing values in the dataset
missing_values = income_df.isnull().sum()
print("Missing values in each column:")
print(missing_values)
```

```python
# Plot a histogram of age with a bin size of 20
plt.figure(figsize=(8, 6))
plt.hist(income_df['age'], bins=20, edgecolor='black')
plt.title('Histogram of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Function to strip the whitespace characters from string values in the dataset
def strip_whitespace(df):
	for column in df.columns:
		if df[column].dtype == 'object':  # Check if the column is of string type
			df[column] = df[column].str.strip()  # Strip whitespace from string values
	return df
```

```python
# Apply strip_whitespace function
for column in income_df.columns:
	if income_df[column].dtype == 'object':  # Check if the column is of string type
		new_column_name = f'{column}_stripped'  # Create a new column name
		income_df[new_column_name] = income_df[column].str.strip()  # Strip whitespace and store in new column
		income_df[column] = income_df[new_column_name]  # Copy values from new column to old column
		income_df.drop(columns=[new_column_name], inplace=True)  # Drop the new column
# Display the first 10 rows of the updated dataset to verify that whitespace has been stripped
income_df.head(10)
```

```python
# Create a DataFrame with only age, education, and occupation by using subsetting
subset_df = income_df[['age', 'education', 'occupation']]
# Display the first 10 rows of the subset DataFrame
subset_df.head(10)
```

```python
# Find the number of people who are aged between 30 and 50
age_30_50_count = ((subset_df['age'] >= 30) & (subset_df['age'] <= 50)).sum()
print(f'Number of people aged between 30 and 50: {age_30_50_count}')

# Textbook example demonstrates "find" actually means generate a new dataframe with the results of the query, so we will create a new DataFrame with only the records of people aged between 30 and 50
age_30_50_df = subset_df[(subset_df['age']>=30) & (subset_df['age']<=50)]
age_30_50_df.head(10)
```

```python
# Group the records based on age and education and find how the mean age is distributed
age_education_group = subset_df.groupby(['age', 'education'])['age'].mean().reset_index(name='mean_age')
age_education_group.head(10)
```

```python
# Group by occupation and show the summary statistics of age
occupation_age_stats = subset_df.groupby('occupation').describe()['age']
print("Summary statistics of age by occupation:")
occupation_age_stats
```

```python
# Find which profession has the oldest workers on average
oldest_profession = occupation_age_stats['mean'].idxmax()
print(f'The profession with the oldest workers on average is {oldest_profession} with an average age of {occupation_age_stats["mean"].max():.2f} years.')

# Find which profession has its largest share of the workforce above the 75th percentile
threshold = subset_df['age'].quantile(0.75)
occupation_above_threshold = subset_df[subset_df['age'] > threshold].groupby('occupation').size()
largest_share_profession = occupation_above_threshold.idxmax()
print(f'The profession with the largest share of workers above the 75th percentile is {largest_share_profession} with a share of {occupation_above_threshold.max() / subset_df.shape[0] * 100:.2f}% of the workforce above the 75th percentile.')
```

```python
# Plot the outliers
plt.figure(figsize=(15,8))
plt.barh(y=occupation_age_stats.index,
         width=occupation_age_stats['count'])
plt.yticks(fontsize=13)
plt.show()

```

```python
# Create DataFrame with only age, workclass, and occupation by using subsetting
age_work_occupation_df = income_df[['age','workclass','occupation']].sample(5,random_state=101)
age_work_occupation_df.head()
```

```python
# Create DataFrame with only education and occupation by using subsetting
education_occupation_df = income_df[['education', 'occupation']].sample(5, random_state=101)
education_occupation_df.head()
```

```python
# Merge on occupation and drop dupes
merged_df = pd.merge(age_work_occupation_df, education_occupation_df, on='occupation', how='inner')
merged_df.drop_duplicates(inplace=True)
merged_df
```

```python
# Create Series 1
series1 = pd.Series([7.3, -2.5, 3.4, 1.5], index=['a', 'b', 'c', 'd'])
series1
```

```python
# Create Series 2
series2 = pd.Series([-2.1, 3.6, -1.5, 4, 3.1], index=['a', 'c', 'e', 'f', 'g'])
series2
```

```python
# Add Series 1 and Series 2
added_series = series1 + series2
print("Result of adding Series 1 and Series 2:")
added_series
```

```python
# Subtract Series 1 from Series 2
subtracted_series = series2 - series1
print("Result of subtracting Series 1 from Series 2:")
subtracted_series
```
