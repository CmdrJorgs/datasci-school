## Activity 8.01

The goal of this activity is to fetch data and retrieve information from two tables, persons and pets, which are a part of the petsdb database.

In this activity, you will first connect to the petsdb database (`./weeks_11_data/petsdb`) and check whether the connection has been successful. You will then create and execute a few SQL commands to answer the following questions: 

- What is the count of people belonging to different age groups in the persons table? 
- Which age group has the maximum number of people? 
- How many people do not have a last name? 
- How many people have more than one pet? 
- How many pets have received treatment? 
- How many pets have received treatment, and the type of pet is known? 
- How many pets are from the city called east port? 
- How many pets are from the city called east port, and who received treatment? 

```python
# Import libraries
import sqlite3

# Connect to the database
conn = sqlite3.connect('./weeks_11_data/petsdb')
cur = conn.cursor()

if conn:
    print("Connection to petsdb successful")
```

```python
# Count of people belonging to different age groups
print("Count of people by age group:")
for row in cur.execute("SELECT age, COUNT(age) FROM persons GROUP BY age"):
    print(row)
```

```python
# Age group with the maximum number of people
cur.execute("SELECT age, COUNT(age) AS count FROM persons GROUP BY age ORDER BY count DESC LIMIT 1")
print("Age group with most people:", cur.fetchone())
```

```python
# How many people do not have a last name
cur.execute("SELECT COUNT(*) FROM persons WHERE last_name IS NULL")
print("People with no last name:", cur.fetchone()[0])
```

```python
# How many people have more than one pet
cur.execute("SELECT COUNT(*) FROM (SELECT owner_id FROM pets GROUP BY owner_id HAVING COUNT(*) > 1)")
print("People with more than one pet:", cur.fetchone()[0])
```

```python
# How many pets have received treatment
cur.execute("SELECT COUNT(*) FROM pets WHERE treatment_done = 1")
print("Pets that received treatment:", cur.fetchone()[0])
```

```python
# How many pets have received treatment and pet type is known
cur.execute("SELECT COUNT(*) FROM pets WHERE treatment_done = 1 AND pet_type IS NOT NULL")
print("Pets treated with known type:", cur.fetchone()[0])
```

```python
# How many pets are from the city called east port
cur.execute("""
    SELECT COUNT(*) FROM pets
    JOIN persons ON pets.owner_id = persons.id
    WHERE persons.city = 'east port'
""")
print("Pets from east port:", cur.fetchone()[0])
```

```python
# How many pets are from east port and received treatment
cur.execute("""
    SELECT COUNT(*) FROM pets
    JOIN persons ON pets.owner_id = persons.id
    WHERE persons.city = 'east port' AND pets.treatment_done = 1
""")
print("Pets from east port that received treatment:", cur.fetchone()[0])
```
