## 1.01

1. Create a list of 100 random numbers. 
2. Create a new list from this random list, with numbers that are divisible by 3. 
3. Calculate the length of these two lists and store the difference in a new variable. 
4. Using a loop, perform steps 1, 2, and 3, and find the difference variable 10 times. 
5. Find the arithmetic mean of these 10 difference values.

```python
# Generate a list of 100 random whole numbers between 1 and 1000
random_numbers_100 = np.random.randint(1, 1001, size=100)
print(f"100 random numbers: {random_numbers_100}")

# Create a new list from the original list containing only the numbers that are divisible by 3
divisible_by_3 = [num for num in random_numbers_100 if num % 3 == 0]
print(f"Numbers divisible by 3: {divisible_by_3}")

# Calculate the length of each list
print(f"Length of random_numbers_100: {len(random_numbers_100)}")
print(f"Length of divisible_by_3: {len(divisible_by_3)}")

# Store length difference in a variable
length_difference = len(random_numbers_100) - len(divisible_by_3)
print(f"Length difference: {length_difference}")

# Loop through above steps 10 times
length_differences = []
for i in range(10):
	random_numbers_100 = np.random.randint(1, 1001, size=100)
	divisible_by_3 = [num for num in random_numbers_100 if num % 3 == 0]
	length_difference = len(random_numbers_100) - len(divisible_by_3)
	print(f"Iteration {i+1}: Length difference: {length_difference}")
	length_differences.append(length_difference)
 
 # Find mean of length differences
mean_length_difference = np.mean(length_differences)
print(f"Mean length difference: {mean_length_difference}")
```

1. Create a mutliline_text variable by copying the text from the first chapter of Pride and Prejudice. Hint: Remember to add triple quotation marks to enter a multiline text. Note: Part of the first chapter of Pride and Prejudice by Jane Austen has been made available on this book's GitHub repository at https://packt.live/2N6ZGP6. 
2. Find the type and length of the multiline_text string using the type and len commands. 
3. Remove all new lines and symbols using the replace method. 
4. Find all of the words in multiline_text using the split method. 
5. Create a list from this list that will contain only the unique words. 
6. Count the number of times the unique word has appeared in the list using the key and value in dict. 
7. Find the top 25 words from the unique words that you have found using the slicemethod. 

```python
# Import P&P Ch01 text file and read it into multiline_text variable
with open('P&P Ch01', 'r') as file:
	multiline_text = file.read()

# Find type and length of multiline_text variable
print(f"Type of multiline_text: {type(multiline_text)}")
print(f"Length of multiline_text: {len(multiline_text)}")

# Remove all new lines and symbols using replace
cleaned_text = multiline_text.replace('\n', '').replace('.', '').replace(',', '').replace('!', '').replace('?', '')
print(f"Cleaned text: {cleaned_text[:100]}...") # Print first 100 characters of cleaned text

# Split cleaned text into a list of words
words_list = cleaned_text.split()
print(f"List of words: {words_list[:10]}...") # Print first 10 words

# Create list that contains only unique words
unique_words = list(set(words_list))
print(f"Unique words: {unique_words[:10]}...") # Print first 10 unique words

# Create dict that counts word frequency
word_counts = {word: words_list.count(word) for word in unique_words}
print(f"Word counts: {list(word_counts.items())[:10]}...") # Print counts of first 10 unique words

# Find top 25 words using slice
top_25_words = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)[:25]
print(f"Top 25 words: {top_25_words}")
```

## 2.01

1. Look up the definition of permutations and dropwhile from itertools. 
2. Write an expression to generate all the possible three-digit numbers, using 0, 1, and 2. 
3. Loop over the iterator expression you generated before. Print each element returned by the iterator. Use assert and isinstance to make sure that the elements are of the tuple type. 
4. Write the loop again, using dropwhile, with a lambda expression to drop any leading zeros from the tuples. As an example, (0, 1, 2) will become [0, 2]. Also, cast the output of dropwhile to a list. 
5. Check the actual type that dropwhile returns. 
6. Combine the preceding code into one block; this time, write a separate function where you will pass the list generated from dropwhile and the function will return the whole number contained in the list. As an example, if you pass [1, 2] to the function, it will return 12. Make sure that the return type is indeed a number and not a string. Although this task can be achieved using other tricks, treat the incoming list as a stack in the function and generate the number by reading the individual digits from the stack.

```python
# Print definition of permutations from itertools
from itertools import permutations
print(f"Permutations: {permutations.__doc__}\n")

# Print definition of dropwhile from itertools
from itertools import dropwhile
print(f"Dropwhile: {dropwhile.__doc__}\n")

# Generate all possible 3 digit numbers using digits 0-2
digits = ['0', '1', '2']
three_digit_numbers = [''.join(p) for p in permutations(digits, 3)]
print(f"All possible 3 digit numbers using digits 0-2: {three_digit_numbers}")

# Loop over iterator
for number in three_digit_numbers:
	print(number)
	assert isinstance(number, str), "Element is not a string" # Check if each element is a string
 
# Loop using dropwhile using lambda to strip leading zeros, and cast the output to a list
non_zero_numbers = [num.lstrip('0') or '0' for num in three_digit_numbers]
print(f"3 digit numbers without leading zeros: {non_zero_numbers}")	

# Combine above steps into one block
# Function that returns whole number from dropwhile list
def generate_number(digit_list):
	number_str = ''
	while digit_list:
		digit_str = digit_list.pop(0) # Treat list as a stack
		number_str += digit_str
	number = int(number_str)
	assert isinstance(number, int), "Output is not an integer" # Check if output is an integer
	return number

# Loop through non_zero_numbers, convert each to a list of digits, and generate the whole number using the function
final_numbers = []
for number_str in non_zero_numbers:
	digit_list = list(number_str)
	final_number = generate_number(digit_list)
	final_numbers.append(final_number)
print(f"Final numbers: {final_numbers}")
```

## 2.02

1. Import zip_longest from itertools. Create a function to zip header, line, and fillvalue=None. Open the accompanying sales_record.csv file from the GitHub link (https://packt.live/2Yb6iCh) by using r mode inside a with block and check that it is opened. 
2. Read the first line and use string methods to generate a list of all the column names. 
3. Start reading the file. Read it line by line. 
4. Read each line and pass that line to a function, along with the list of the headers. The work of the function is to construct a dictionary out of these two and fill up the key:values variables. Keep in mind that a missing value should result in None.

```python
# Import zip_longest from itertools
from itertools import zip_longest

# Define function to zip header, line, and fillvalue=None
def zip_with_fill(header, line, fillvalue=None):
	zipped = zip_longest(header, line, fillvalue=fillvalue)
	return list(zipped)

# Import sales_record.csv using r mode inside a with block and check that it is opened
with open('sales_record.csv', 'r') as file:
	sales_data = file.read()
	print("File opened successfully")
 
# Read first line and use string method to generate list of all column names
lines = sales_data.splitlines()
header = lines[0].split(',')
print(f"Column names: {header}")

# Define function to construct a dictionary out of header list and line list into key:value pairs (missing values return key:None)
def construct_dict(header, line):
	return dict(zip_with_fill(header, line, fillvalue=None))

# Loop through each line and pass the line toa function along with header list
data_dicts = []
for line in lines[1:]:
	line_list = line.split(',')
	data_dict = construct_dict(header, line_list)
	data_dicts.append(data_dict)
print(f"Data dictionaries: {data_dicts[:2]}") # Print first 2 dictionaries
```
